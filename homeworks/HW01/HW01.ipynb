# Примеры векторных операций и базовых статистик:

```python
# ax^2 + bx + c = 0
x = np.array([6,2,5,1,7])
y = np.array([52,16,40,12,66])

# МНК для полинома 2-й степени
# Обучение модели
A = np.array([[np.float64(len(x)), sum(x), sum(x**2)],
       [sum(x), sum(x**2), sum(x**3)],
       [sum(x**2), sum(x**3), sum(x**4)]],dtype=float)

B = np.array([sum(y),sum(y*x),sum(x**2*y)]).reshape((3,1))

weights = np.linalg.solve(A,B).ravel()
print('Веса модели (Коэффициенты полиномиальной регрессии): ', weights)

y_pred = x**2*weights[2]+ x*weights[1] + weights[0] 

MAE = sum(y - y_pred) / len(x)
print('Средняя абсолютная ошибка:', MAE)
```

Попытка №2. Примеры операций с массивами, простые вычисления/статистик (mean/sum и т.п)

```python
u = np.array([1,2,1,2,2,1,1])
umean = np.mean(u)
usum = np.sum(u)
umedian = np.median(u)
print(f'mean u: {umean}')
print(f'sum u: {usum}')
print(f'median u: {umedian}')
```
